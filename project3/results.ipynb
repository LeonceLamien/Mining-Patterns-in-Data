```python
import numpy as np
import matplotlib.pyplot as plt

# Get data for each method
bif_results = [alarm_BIF, andes_BIF, asia_BIF, sachs_BIF, sprinkler_BIF, water_BIF]
param_results = [alarm_param, andes_param, asia_param, sachs_param, sprinkler_param, water_param]
struct_results = [alarm_struct, andes_struct, asia_struct, sachs_struct, sprinkler_struct, water_struct]

# Set width of bars and positions
bar_width = 0.25
x = np.arange(6)

# Create the figure and axes
fig, ax = plt.subplots(figsize=(15, 5))

# Create bars with different blue shades
bars1 = ax.bar(x - bar_width, bif_results, bar_width, label='Pre-computed CPTs', color='#1f77b4')  # dark blue
bars2 = ax.bar(x, param_results, bar_width, label='Parameter Learning', color='#7db5d6')  # medium blue
bars3 = ax.bar(x + bar_width, struct_results, bar_width, label='Structure Learning', color='#c6dbef')  # light blue

# Add labels, title and legend
ax.set_xlabel('Datasets')
ax.set_ylabel('Accuracy')
ax.set_title('Comparison of Methods Across Different Datasets')
ax.set_xticks(x)
ax.set_xticklabels(['Alarm', 'Andes', 'Asia', 'Sachs', 'Sprinkler', 'Water'])
ax.legend(bbox_to_anchor=(1.05, 1), loc='upper left')

# Add zero markers for None values
for i, val in enumerate(struct_results):
    if val == 0:
        ax.text(i + bar_width, 0.05, 'None', ha='center', rotation=90)

# Add light gray grid lines for accuracy values
ax.yaxis.grid(True, linestyle='--', linewidth=0.7, alpha=0.7, color='lightgray')

# Remove the border
ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)
ax.spines['bottom'].set_visible(False)
ax.spines['left'].set_visible(False)

plt.tight_layout()
plt.show()
```
